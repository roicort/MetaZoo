```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

from metazoo.bio.evolutionary import GeneticProgramming
from metazoo.bio.evolutionary.gp import Operators, OperatorSet, Tree

import plotly.io as pio
pio.renderers.default = "jupyterlab"
#pio.renderers.default = "svg"
```

```python
Operators.available()
```

```python
op = OperatorSet()
op.add_function( Operators.AND )
op.add_function( Operators.OR )
op.add_function( Operators.XOR )
op.add_terminal( 0 )
op.add_terminal( 1 )
op.resume()
```

```python
tree = Tree(operator_set=op)
tree.generate(max_depth=5, method='ramped_half_and_half')
fig = tree.plot()
fig.show()
```

```python
print(str(tree))
print(tree.eval())
```

```python
gp = GeneticProgramming(
    fitness_function=lambda x: np.sum(x, axis=1),
    operator_set=op, 
    population_size=100
)

gp.summary()
```

```python
gp.run(generations=1, verbose=False)
```
